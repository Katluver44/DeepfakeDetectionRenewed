# In-The-Wild — 30 random clips player
Loads 30 random samples from the local In-The-Wild dataset and gives an audio widget for each.

```python
import numpy as np
from pathlib import Path
from IPython.display import display, Audio, HTML
from datasets import load_dataset

CACHE_DIR   = "data/in_the_wild"
N_SAMPLES   = 30
RANDOM_SEED = 42  # change to get a different set

print("Loading dataset (cached locally)...")
ds = load_dataset("mueller91/In-The-Wild", cache_dir=CACHE_DIR, trust_remote_code=True)["train"]
print(f"Dataset size: {len(ds):,} clips")

rng  = np.random.default_rng(RANDOM_SEED)
idxs = sorted(rng.choice(len(ds), N_SAMPLES, replace=False).tolist())
print(f"Sampled indices: {idxs}")
```

```python
# Load all 30 samples (decodes audio from the .wav files)
samples = [ds[i] for i in idxs]
print(f"Loaded {len(samples)} clips.")
```

```python
# Display all 30 clips with playable widgets
for rank, (i, s) in enumerate(zip(idxs, samples), 1):
    audio   = s["audio"]
    arr     = np.array(audio["array"], dtype=np.float32)
    sr      = audio["sampling_rate"]
    fname   = Path(audio["path"]).name
    dur_sec = len(arr) / sr

    display(HTML(
        f"<b>#{rank:02d}</b> &nbsp; idx={i} &nbsp; file={fname} &nbsp; "
        f"sr={sr} Hz &nbsp; dur={dur_sec:.2f}s"
    ))
    display(Audio(data=arr, rate=sr))
```

```python
# ── Resample with a different seed ──────────────────────────────────────────
# Edit NEW_SEED below and re-run this cell + the two above to hear a fresh set.

NEW_SEED = 123

rng2   = np.random.default_rng(NEW_SEED)
idxs2  = sorted(rng2.choice(len(ds), N_SAMPLES, replace=False).tolist())
samples2 = [ds[i] for i in idxs2]

for rank, (i, s) in enumerate(zip(idxs2, samples2), 1):
    audio   = s["audio"]
    arr     = np.array(audio["array"], dtype=np.float32)
    sr      = audio["sampling_rate"]
    fname   = Path(audio["path"]).name
    dur_sec = len(arr) / sr

    display(HTML(
        f"<b>#{rank:02d}</b> &nbsp; idx={i} &nbsp; file={fname} &nbsp; "
        f"sr={sr} Hz &nbsp; dur={dur_sec:.2f}s"
    ))
    display(Audio(data=arr, rate=sr))
```
